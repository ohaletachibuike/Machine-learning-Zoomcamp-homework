```python
# Q3 Fuel types
fuel_type_list=df['fuel_type'].unique()
print("Number of fuel types are : ", len(fuel_type_list))
```

```python
# Q4 Missing Values
missing_cols=df.isna().sum()
num_missing_cols=(missing_cols>0).sum()
print("Number of missing columns are : ", num_missing_cols)
```

```python
# Q5 Max fuel efficiency
fuel_efficiency_of_asian_cars=df[df['origin']=='Asia'].fuel_efficiency_mpg
print("Maximum fuel efficiency within asian cars is : ",max(fuel_efficiency_of_asian_cars))
```

```python
# Q6 Median Value of horsepower
horsepower_median=df['horsepower'].median()
print("horsepower median before filling : ", horsepower_median)

most_freq_horsepower=df['horsepower'].mode()[0]
df['horsepower'].fillna(most_freq_horsepower)
print("Missing values of horsepower are filled with most frequent value which is : ", most_freq_horsepower)

horsepower_median_new=df['horsepower'].median()
print("horsepower median after filling : ", horsepower_median_new)
```

```python
# Q7 Sum of weights

asian_cars=df[df['origin']=='Asia'][['vehicle_weight','model_year']].head(7)
asian_cars
```

```python
X=np.array(asian_cars)
XTX=np.dot(X.T, X)
inverse_XTX=np.linalg.inv(XTX)
y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w=inverse_XTX.dot(X.T).dot(y)

print("Sum of all the elements of the result w : ", np.sum(w))
```
